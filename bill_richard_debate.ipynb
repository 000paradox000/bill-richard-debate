## Libraries

````python
import ollama
````

````python
from langchain.prompts import ChatPromptTemplate
from langchain_core.runnables import Runnable
from langchain_ollama import ChatOllama
````

## Globant variables

````python
with open("bill-richard-debate/richard_stallman.txt", "r") as fh:
    PROMPT_RICHARD = fh.read().strip()
````

````python
with open("bill-richard-debate/bill_gates.txt", "r") as fh:
    PROMPT_BILL = fh.read().strip()
````

````python
MODEL_NAME = "gemma3:1b"
````

````python
QUERIES = [
    "¿Cómo la tecnología contribuye a la democratización del acceso a la información y la educación?",
]
````

````python
LLM = ChatOllama(model=MODEL_NAME, temperature=0.7)
````

````python
AGENT_RICHARD =ChatPromptTemplate.from_template(PROMPT_RICHARD) | LLM
````

````python
AGENT_BILL =ChatPromptTemplate.from_template(PROMPT_BILL) | LLM
````

## Debate

````python
turns = 6
agents = [
    ("[BILL]", AGENT_BILL),
    ("[RICHARD]", AGENT_RICHARD),
]

for query in QUERIES:
    print_query = f"💬 Conversation started: {query}"
    log_lines = [print_query]
    print(print_query)
    msg = query

    for i in range(turns):
        label, agent = agents[i % len(agents)]
        msg = agent.invoke({"input": msg}).content
        entry = f"{label}: {msg}"
        print(f"\n{entry}")
        log_lines.append(entry)

with open("chat.md", "w", encoding="utf-8") as fh:
    fh.write("\n\n".join(log_lines))

print("\n📝 Conversation saved")
````

````python
from gtts import gTTS
from IPython.display import Audio #Import Audio method from IPython's Display Class
````

````python
with open("chat.md", "r") as fh:
    contents = fh.read()
````

````python
import re
from collections import defaultdict

def extraer_dialogo_por_personaje(texto: str) -> dict[str, str]:
    # Diccionario para guardar el texto hablado por cada personaje
    dialogo = defaultdict(str)

    # Regex para encontrar cada bloque de intervención
    bloques = re.split(r'\[(\w+)\]:', texto)

    # Alterna entre nombre e intervención
    for i in range(1, len(bloques), 2):
        personaje = bloques[i].strip()
        texto_personaje = bloques[i + 1].strip()
        dialogo[personaje] += texto_personaje + '\n\n'

    return dict(dialogo)

# O como string directamente
contenido = contents

resultado = extraer_dialogo_por_personaje(contenido)

# Mostrar resultados por pantalla o guardarlos en archivos separados
for personaje, habla in resultado.items():
    print(f"--- {personaje} ---\n")
    print(habla)

    # Si prefieres guardar cada parte en un archivo:
    with open(f"{personaje}.txt", "w", encoding="utf-8") as f:
        f.write(habla)
````

````python
tts = gTTS(contents, lang="es") #Provide the string to convert to speech
tts.save('1.wav') #save the string converted to speech as a .wav file
sound_file = '1.wav'
Audio(sound_file, autoplay=True)
````
